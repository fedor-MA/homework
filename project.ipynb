```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Загрузка
df = pd.read_excel('Road Accident Data.xlsx')


print(df.shape)           # (сколько строк, сколько столбцов)
print(df.columns)         # какие признаки есть
print(df.info())          # типы данных (числа, текст, даты)
print(df.head())          # первые 5 строк
print(df.tail())          # последние 5 строк

print(df.isnull().sum())                    # сколько пропусков в каждом столбце
print(df.isnull().sum() / len(df) * 100)    # процент пропусков

# Дубликаты
print(df.duplicated().sum())                # есть ли полностью одинаковые строки

# Уникальные значения в категориальных признаках
print(df['Accident_Severity'].value_counts())

# Сколько аварий каждого уровня тяжести
severity_counts = df['Accident_Severity'].value_counts()
print(severity_counts)
print(severity_counts / len(df) * 100)  # в процентах

# График (обязательно с подписями!)
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Accident_Severity', palette='Reds')
plt.title('Распределение аварий по уровню тяжести', fontsize=14)
plt.xlabel('Уровень тяжести (Severity)', fontsize=12)
plt.ylabel('Количество аварий', fontsize=12)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

df = pd.read_excel('Road Accident Data.xlsx')


print(f"\n1. Размер датасета: {df.shape[0]} строк, {df.shape[1]} столбцов")
print(f"\n2. Столбцы: {df.columns.tolist()}")
print(f"\n3. Типы данных:")
print(df.dtypes)

print(f"\n4. Пропуски в данных:")
print(df.isnull().sum())
print(f"\nПроцент пропусков:")
print((df.isnull().sum() / len(df) * 100).round(2))

print(f"\n5. Дубликаты: {df.duplicated().sum()}")

print(f"\n6. Распределение Accident_Severity:")
severity_counts = df['Accident_Severity'].value_counts()
print(severity_counts)
print(f"\nВ процентах:")
print((severity_counts / len(df) * 100).round(2))

plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Accident_Severity', palette='Reds')
plt.title('Распределение аварий по уровню тяжести', fontsize=14, fontweight='bold')
plt.xlabel('Уровень тяжести (Accident Severity)', fontsize=12)
plt.ylabel('Количество аварий', fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Значения нахдятся в ключевых столбцах
print(f"\n7. Уникальные значения в ключевых признаках:")
print(f"Weather_Conditions: {df['Weather_Conditions'].unique()}")
print(f"Road_Surface_Conditions: {df['Road_Surface_Conditions'].unique()}")
print(f"Light_Conditions: {df['Light_Conditions'].unique()}")
print(f"Road_Type: {df['Road_Type'].unique()}")
print(f"Urban_or_Rural_Area: {df['Urban_or_Rural_Area'].unique()}")

# Связь погоды с тяжестью аварии
plt.figure(figsize=(12,6))
weather_severity = pd.crosstab(df['Weather_Conditions'], df['Accident_Severity'], normalize='index') * 100
weather_severity.plot(kind='bar', stacked=True, figsize=(12,6), colormap='Reds')
plt.title('Тяжесть аварии в зависимости от погодных условий', fontsize=14, fontweight='bold')
plt.xlabel('Погодные условия', fontsize=12)
plt.ylabel('Процент аварий', fontsize=12)
plt.legend(title='Accident Severity', bbox_to_anchor=(1.05, 1))
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Связь состояния дороги с тяжестью аварии
plt.figure(figsize=(10,6))
sns.countplot(data=df, x='Road_Surface_Conditions', hue='Accident_Severity')
plt.title('Тяжесть аварии в зависимости от состояния дороги', fontsize=14, fontweight='bold')
plt.xlabel('Состояние дороги', fontsize=12)
plt.ylabel('Количество аварий', fontsize=12)
plt.legend(title='Accident Severity')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

df['Accident_Date'] = pd.to_datetime(df['Accident Date'])

# Аварии по дням недели
plt.figure(figsize=(10,5))
sns.countplot(data=df, x='Day_of_Week', hue='Accident_Severity')
plt.title('Количество аварий по дням недели', fontsize=14, fontweight='bold')
plt.xlabel('День недели', fontsize=12)
plt.ylabel('Количество аварий', fontsize=12)
plt.legend(title='Accident Severity')
plt.tight_layout()
plt.show()

# Анализ числовых признаков
numeric_cols = ['Number_of_Casualties', 'Number_of_Vehicles', 'Speed_limit']

for col in numeric_cols:
    fig, axes = plt.subplots(1, 2, figsize=(12,4))
    
    sns.boxplot(data=df, y=col, ax=axes[0])
    axes[0].set_title(f'Boxplot: {col}', fontweight='bold')
    axes[0].set_ylabel(col)
    
    sns.histplot(data=df, x=col, bins=30, ax=axes[1], kde=True)
    axes[1].set_title(f'Гистограмма: {col}', fontweight='bold')
    axes[1].set_xlabel(col)
    
    plt.tight_layout()
    plt.show()
    
    print(f"\nСтатистика для {col}:")
    print(df[col].describe())

# Связь количества машин с тяжестью аварии
plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Accident_Severity', y='Number_of_Vehicles')
plt.title('Количество транспортных средств vs Тяжесть аварии', fontsize=14, fontweight='bold')
plt.xlabel('Уровень тяжести аварии', fontsize=12)
plt.ylabel('Количество ТС', fontsize=12)
plt.tight_layout()
plt.show()
```
